```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("Expanded_data_with_more_features.csv")
```

```python
if 'Unnamed: 0' in df.columns:
    df.drop('Unnamed: 0', axis=1, inplace=True)

df['NrSiblings'] = df['NrSiblings'].fillna(df['NrSiblings'].median())
```

```python

categorical_cols = ['EthnicGroup', 'ParentEduc', 'TestPrep', 'ParentMaritalStatus', 
                    'PracticeSport', 'IsFirstChild', 'TransportMeans', 'WklyStudyHours']

for col in categorical_cols:
    df[col] = df[col].fillna(df[col].mode()[0])

# (Mapping/Encoding)
study_mapping = {'< 5': 1, '5 - 10': 2, '> 10': 3}
value_mapping = {'no': 0, 'yes': 1}
test_mapping = {'none': 0, 'completed': 1}

df['WklyStudyHours_Encoded'] = df['WklyStudyHours'].map(study_mapping)
df['IsFirstChild_Encoded'] = df['IsFirstChild'].map(value_mapping)
df['TestPrep_Encoded'] = df['TestPrep'].map(test_mapping)
```

```python
print("\nعدد القيم المفقودة بعد التنظيف")
print(df.isnull().sum())
```

```python
from IPython.display import display

print("\nUsing loc:")
display(df.loc[0:4, ['Gender', 'MathScore', 'ReadingScore']])

print("\nUsing iloc:")
display(df.iloc[0:5, 0:3])
```

```python

# 2. (Multi-condition Filtering)
filtered_data = df[(df['MathScore'] > 80) & (df['TestPrep'] == 'completed')]
print("\nFiltered Data (MathScore > 80 AND TestPrep = completed):", len(filtered_data))

# 3. (Sorting)
sorted_data = df.sort_values(by='MathScore', ascending=False)
print("\nTop 5 Students in Math:")
print(sorted_data[['Gender', 'MathScore']].head())

# 4.  (GroupBy)
grouped_data = df.groupby('ParentEduc')[['MathScore', 'ReadingScore', 'WritingScore']].mean()
print("\nAverage Scores by Parent Education:")
print(grouped_data)
```

```python
plt.figure(figsize=(15, 12))

# Plot 1: Histogram / KDE (Math Score Distribution)
plt.subplot(3, 2, 1)
sns.histplot(df['MathScore'], kde=True, color='blue')
plt.title('1. Distribution of Math Scores')

# Plot 2: Boxplot (Math Score by Gender)
plt.subplot(3, 2, 2)
sns.boxplot(x='Gender', y='MathScore', data=df, hue='Gender', palette='Set2', legend=False)
plt.title('2. Math Scores by Gender')

# Plot 3: Bar Chart (Average Math Score by Parent Education)
plt.subplot(3, 2, 3)
sns.barplot(x='MathScore', y='ParentEduc', data=df, errorbar=None, hue='ParentEduc', palette='viridis', legend=False)
plt.title('3. Mean Math Score by Parent Education')

# Plot 4: Scatter Plot (Math Score vs Reading Score)
plt.subplot(3, 2, 4)
sns.scatterplot(x='MathScore', y='ReadingScore', hue='TestPrep', data=df, alpha=0.5)
plt.title('4. Math Score vs Reading Score')

# Plot 5: Correlation Heatmap
plt.subplot(3, 2, 5)
sns.heatmap(df[['MathScore', 'ReadingScore', 'WritingScore', 'NrSiblings']].corr(), annot=True, cmap='coolwarm', fmt='.2f')
plt.title('5. Correlation Matrix Heatmap')

plt.tight_layout()
plt.show()
```
